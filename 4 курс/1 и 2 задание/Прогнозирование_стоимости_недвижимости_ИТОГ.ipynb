## Описание набора данных

**Источник данных:** платформа Zingat — турецкий онлайн-сервис объявлений
о недвижимости, выступающий посредником между покупателями и продавцами.
Датасет содержит информацию об объектах недвижимости, выставленных на
продажу и в аренду.

**Целевая переменная:** `price` — стоимость объекта недвижимости.

**Признаки:**

- **type** (Тип недвижимости): категориальный признак. Значение `Konut`
  соответствует жилой недвижимости.
- **sub_type** (Подтип недвижимости): категориальный признак. Значение
  `Daire` соответствует квартире.
- **start_date** (Дата публикации): дата, когда объявление стало активным
  на рынке.
- **end_date** (Дата снятия): дата, когда объявление перестало быть
  активным.
- **listing_type** (Тип листинга): категориальный признак с двумя
  значениями — `Satılık` (продажа) и `Kiralık` (аренда).
- **tom** (Time on market): время нахождения объекта на рынке.
- **building_age** (Возраст здания): возраст постройки в годах.
  Значение `arası` означает «между» (диапазон).
- **total_floor_count** (Этажность): общее количество этажей в здании.
- **room_count** (Количество комнат): например, `2+1` означает
  2 комнаты + 1 гостиная.
- **size** (Площадь): площадь объекта в м².

# Этап 2. Предварительная обработка данных 

## 2.1 Работа с пропусками

### Первичная очистка данных

Перед непосредственной обработкой пропущенных значений выполняется
первичная очистка исходного набора данных.

```python
df_clean = df.copy()

# Удаляем признаки, которые не будем использовать в модели
df_clean.drop(columns=['furnished', 'type'], inplace=True)

# Удаляем полные дубликаты
duplicates = df_clean.duplicated().sum()
print(f"Количество полных дубликатов: {duplicates:,}")

df_clean = df_clean.drop_duplicates().copy()

# Для модели рассматриваем только объявления о продаже
df_clean = df_clean[df_clean['listing_type'] == 1].copy()

# Оставляем только цены в TRY
df_clean = df_clean[df_clean['price_currency'] == 'TRY'].copy()

# Удаляем строки без цены
df_clean = df_clean.dropna(subset=['price']).copy()

# Цена не может быть отрицательной или нулевой
df_clean = df_clean[df_clean['price'] > 0].copy()

print(f"Размер данных после очистки: {df_clean.shape}")
```

## 2.2 Обработка новой

```python
df_proc = df_clean.copy()
```

### Винзоризация

```python
price_lower = df_proc['price'].quantile(0.01)
price_upper = df_proc['price'].quantile(0.99)

print(f"Цена: от {price_lower:,.0f} до {price_upper:,.0f} (1-99 перцентили)")
```

```python
df_proc = df_proc[(df_proc['price'] >= price_lower) & (df_proc['price'] <= price_upper)]
```

На этапе обработки выбросов я применила винзоризацию по целевой переменной — отсекла 1% самых дешёвых и 1% самых дорогих объектов. Диапазон цен сузился с исходного до 65 000–3 600 000 лир, что позволило убрать экстремальные выбросы, не теряя при этом значимую часть данных — мы потеряли лишь около 4% выборки. 

### Логарифмирование сильно скошенных признаков

```python
df_proc['log_price'] = np.log1p(df_proc['price'])
df_proc['log_size'] = np.log1p(df_proc['size'])
```

```python
df_proc.shape
```

Дополнительно я создала логарифмированные версии цены и площади через np.log1p. Это стандартный приём для работы с правосторонне скошенными распределениями: логарифм сжимает "длинный хвост" и делает распределение более симметричным, что особенно важно для линейных моделей, чувствительных к выбросам. ## 2.3 Кодирование категориальных признаков 

```python
X_sale_enc = X_sale.copy()
X_sale_enc.shape
```

```python
# Адрес делим на город и район
X_sale_enc['city'] = X_sale_enc['address'].str.split('/').str[0].str.strip()
X_sale_enc['district'] = X_sale_enc['address'].str.split('/').str[1].str.strip()

X_sale_enc['district'] = X_sale_enc['district'].fillna('Unknown')
```

```python
# Комнаты извлекаем количество комнат и гостиных
X_sale_enc['rooms'] = X_sale_enc['room_count'].str.extract(r'^(\d+)').astype(float)
X_sale_enc['living_rooms'] = X_sale_enc['room_count'].str.extract(r'\+(\d+)').astype(float).fillna(0)
```

```python
# Возраст здания
X_sale_enc['building_age_num'] = X_sale_enc['building_age'].str.extract(r'(\d+)').astype(float).fillna(10)
```

```python
# Количество этажей
X_sale_enc['total_floors_num'] = X_sale_enc['total_floor_count'].str.extract(r'^(\d+)').astype(float).fillna(5)
```

#### Извлечение признаков из текста (Feature Extraction)

```python
X_sale_enc['is_penthouse'] = X_sale_enc['floor_no'].str.contains('Çatı|En Üst', na=False).astype(int)
X_sale_enc['is_garden'] = X_sale_enc['floor_no'].str.contains('Bahçe', na=False).astype(int)
X_sale_enc['is_ground'] = X_sale_enc['floor_no'].str.contains('Giriş|Zemin|Kot|Yüksek', na=False).astype(int)
# если это просто число
X_sale_enc['floor_num'] = X_sale_enc['floor_no'].str.extract(r'^(\d+)').astype(float)
X_sale_enc['floor_num'] = X_sale_enc['floor_num'].fillna(0)
```

На этом этапе я превратила текстовые и категориальные признаки в числовые, пригодные для обучения модели. Из столбца с этажом я извлекла четыре новых признака: три бинарных флага для нестандартных уровней — пентхаус, садовый этаж и цокольный/входной уровень — и один числовой признак с номером этажа. Это важно, потому что пентхаусы и цокольные квартиры имеют систематически отличающиеся цены, и модель должна это учитывать.


#### Группировка редких категорий

```python
# sub_type: топ-5 + "Other"
top_subtypes = X_sale_enc['sub_type'].value_counts().head(5).index
X_sale_enc['sub_type_grouped'] = X_sale_enc['sub_type'].apply(lambda x: x if x in top_subtypes else 'Other')

# heating_type: топ-6 + "Other"
top_heating = X_sale_enc['heating_type'].value_counts().head(6).index
X_sale_enc['heating_grouped'] = X_sale_enc['heating_type'].apply(lambda x: x if x in top_heating else 'Other')
```

Далее я сгруппировала редкие категории в признаках типа недвижимости и типа отопления: оставила только топ-5 и топ-6 наиболее частых категорий соответственно, а всё остальное объединила в категорию "Other". Это снижает размерность и предотвращает переобучение на редких классах.


### One-Hot Encoding

```python
cols_to_encode = ['sub_type_grouped', 'heating_grouped', 'price_currency', 'city']
X_sale_enc = pd.get_dummies(X_sale_enc, columns=cols_to_encode, drop_first=True, dtype=int)
```

После этого применила One-Hot Encoding к четырём категориальным признакам — сгруппированным типу недвижимости, типу отопления, валюте и городу — с параметром drop_first=True для избежания мультиколлинеарности.


```python
cols_to_drop = ['address', 'room_count', 'building_age', 'total_floor_count', 
                'floor_no', 'sub_type', 'heating_type', 'start_date', 'listing_date', 'tom']
X_sale_enc.drop(columns=[c for c in cols_to_drop if c in X_sale_enc.columns], inplace=True)

X_sale_enc.shape
```

В завершение удалила все исходные текстовые столбцы, которые были заменены числовыми признаками. Итоговая матрица признаков содержит 271 033 объекта и 102 столбца, полностью готовых к масштабированию и обучению модели.

## 2.4 Работа с Feature Engineering

```python
X_feat = X_sale_enc.copy()

if 'district' in X_feat.columns:
    X_feat.drop(columns=['district'], inplace=True)

X_feat['size_per_room'] = X_feat['size'] / (X_feat['rooms'] + 1)
X_feat['is_new_building'] = (X_feat['building_age_num'] <= 5).astype(int)
X_feat['floor_ratio'] = X_feat['floor_num'] / (X_feat['total_floors_num'] + 1)
X_feat['has_living_room'] = (X_feat['living_rooms'] > 0).astype(int)

corr_matrix = X_feat.select_dtypes(include=['number']).corr()

high_corr = []
for i in range(len(corr_matrix.columns)):
    for j in range(i):
        if abs(corr_matrix.iloc[i, j]) > 0.8:
            high_corr.append((corr_matrix.columns[i], corr_matrix.columns[j], corr_matrix.iloc[i, j]))

cols_to_drop = set()
for col1, col2, val in high_corr:
    if val > 0.85:
        if X_feat[col1].nunique() < X_feat[col2].nunique():
            cols_to_drop.add(col1)
        else:
            cols_to_drop.add(col2)
            
X_feat = X_feat.dropna(subset=['rooms', 'size_per_room'])
print(f"Размер после удаления NaN: {X_feat.shape}")

X_feat = X_feat.drop(columns=['log_price', 'log_size'], errors='ignore')
```

На этом этапе я создала четыре новых информативных признака без утечки данных: площадь на комнату, флаг нового здания, отношение этажа к общему числу этажей и флаг наличия гостиной. Эти признаки помогают модели уловить нелинейные зависимости, которые не видны в исходных данных.

Затем я провела анализ мультиколлинеарности — нашла пары признаков с высокой корреляцией (более 0.8) и удалила менее информативные из пар с корреляцией выше 0.85. Это важно, потому что мультиколлинеарность дестабилизирует линейные модели и раздувает дисперсию оценок.

В завершение удалила строки с пропусками в ключевых признаках и вспомогательные логарифмированные столбцы. Итоговая матрица признаков содержит 173 994 объекта и 105 столбцов, полностью готовых к масштабированию и обучению модели.

## 3.3 Метрики качества

```python
def calc_metrics(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
    r2 = r2_score(y_true, y_pred)
    return {'MAE': mae, 'RMSE': rmse, 'MAPE': mape, 'R2': r2}

models = {
    'Linear Regression': best_ridge,
    'Decision Tree': best_dt,
    'Random Forest': best_rf,
    'Gradient Boosting': best_gb
}

res = {}
for name, model in models.items():
    pred = model.predict(X_test_imp)
    res[name] = calc_metrics(y_test, pred)

display(pd.DataFrame(res).T.round(2))
```

На этом этапе я рассчитала четыре метрики качества для каждой модели на тестовой выборке, которую модель не видела при обучении. Результаты показали, что линейная модель Ridge проигрывает всем древовидным — её R² всего 0.42, что подтверждает вывод из EDA о нелинейной зависимости цены от признаков. Лучшей по R² и RMSE оказалась модель Random Forest с показателем 0.54 — она объясняет больше половины дисперсии цен и наиболее устойчива к выбросам. Interestingly, Decision Tree показала наименьший MAE, но её высокий RMSE говорит о том, что она допускает редкие, но очень крупные ошибки. MAPE у всех моделей высокий (29–49%), что объясняется огромным разбросом цен в датасете — от 50 тысяч до миллионов лир. На дешёвых объектах относительная ошибка всегда получается больше. В целом, R² около 0.54 — это разумный результат для задачи прогнозирования недвижимости, где половина ценообразования зависит от факторов, не представленных в данных: точного расположения, вида, состояния ремонта и инфраструктуры района. Для дальнейшего использования в приложении я выбираю Random Forest как наиболее сбалансированную модель

На предыдущих этапах проекта были исследованы данные о недвижимости, выполнена предварительная обработка признаков и рассмотрены модели машинного обучения для прогнозирования стоимости недвижимости.

На данном этапе необходимо проверить работу модели на отдельном наборе из 20 объектов, предоставленном в файле `test_real_estate_samples.csv`.

Для обучения используются исходные данные `real_estate_data.csv`. Тестовые объекты исключаются из обучающей выборки, чтобы избежать утечки данных.

Поскольку исходный проект посвящён прогнозированию стоимости продажи недвижимости, модель проверяется на объявлениях о продаже в турецких лирах (TRY). Объявления об аренде сохраняются в итоговой таблице, но прогноз для них не рассчитывается.

## 4.1. Подключение библиотек и загрузка данных

На этом этапе подключаются библиотеки, необходимые для обучения модели, обработки признаков и оценки качества прогнозирования.

Загружаются два файла: исходный набор данных, используемый для обучения, и тестовый набор из 20 объектов, для которых необходимо получить прогнозы.

```python
import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestRegressor
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import RobustScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Загрузка исходного датасета и тестовых объектов
raw_full = pd.read_csv("real_estate_data.csv", low_memory=False)
test_samples = pd.read_csv("test_real_estate_samples.csv")

print("Размер исходного датасета:", raw_full.shape)
print("Размер тестовой выборки:", test_samples.shape)

display(test_samples.head())
```

## 4.2. Подготовка обучающей выборки

Перед обучением необходимо исключить тестовые объекты из исходного набора данных по идентификаторам `id`. Это важно, поскольку фактические цены тестовых объектов уже присутствуют в исходном датасете.

Далее оставляются объявления о продаже, для которых цена указана в турецких лирах и является положительным числом. Полные дубликаты удаляются, а крайние 1% значений цены исключаются для уменьшения влияния экстремальных наблюдений.

Так формируется обучающая выборка, которая не содержит проверяемые объекты.

```python
test_ids = set(test_samples["id"])
train = raw_full.drop_duplicates().copy()

# Оставляем объявления о продаже в TRY с положительной ценой
train = train[
    (train["listing_type"] == 1) &
    (train["price_currency"] == "TRY") &
    (train["price"] > 0)
].copy()

# Исключаем тестовые объекты из обучения
train = train[~train["id"].isin(test_ids)].copy()

# Удаляем крайние 1% цен
q_low = train["price"].quantile(0.01)
q_high = train["price"].quantile(0.99)

train = train[train["price"].between(q_low, q_high)].copy()

print("Количество объектов для обучения:", len(train))
print("Минимальная цена после фильтрации:", round(train["price"].min(), 2))
print("Максимальная цена после фильтрации:", round(train["price"].max(), 2))
print("Количество тестовых ID, исключённых из обучения:", len(test_ids.intersection(set(raw_full["id"]))))
```

## 4.3. Создание и преобразование признаков

Для работы модели текстовые характеристики недвижимости преобразуются в числовые признаки.

Из адреса извлекается город. Из описания комнат выделяются количество комнат и гостиных. Из характеристик здания извлекаются возраст и общее количество этажей.

Дополнительно создаются признаки, описывающие тип этажа, площадь на комнату, наличие гостиной и относительное положение этажа в здании.

Категориальные признаки переводятся в числовой формат методом One-Hot Encoding. Такой подход позволяет модели Random Forest использовать информацию о типе недвижимости, отоплении, валюте и городе.

```python
def make_features(data, top_subtypes=None, top_heating=None):
    d = data.copy()

    # Город из адреса
    address_parts = (
        d["address"]
        .fillna("Unknown")
        .astype(str)
        .str.split("/")
    )
    d["city"] = address_parts.str[0].str.strip()

    # Количество комнат и гостиных
    room_text = d["room_count"].fillna("").astype(str)

    d["rooms"] = pd.to_numeric(
        room_text.str.extract(r"^(\d+)")[0],
        errors="coerce"
    )

    d["living_rooms"] = pd.to_numeric(
        room_text.str.extract(r"\+(\d+)")[0],
        errors="coerce"
    ).fillna(0)

    # Возраст здания
    d["building_age_num"] = pd.to_numeric(
        d["building_age"].astype(str).str.extract(r"(\d+)")[0],
        errors="coerce"
    ).fillna(10)

    # Общее количество этажей
    d["total_floors_num"] = pd.to_numeric(
        d["total_floor_count"].astype(str).str.extract(r"^(\d+)")[0],
        errors="coerce"
    ).fillna(5)

    # Информация об этаже
    floor = d["floor_no"].fillna("").astype(str)

    d["is_penthouse"] = floor.str.contains(
        r"Çatı|En Üst", case=False, regex=True
    ).astype(int)

    d["is_garden"] = floor.str.contains(
        "Bahçe", case=False
    ).astype(int)

    d["is_ground"] = floor.str.contains(
        r"Giriş|Zemin|Kot|Yüksek", case=False, regex=True
    ).astype(int)

    d["floor_num"] = pd.to_numeric(
        floor.str.extract(r"^(\d+)")[0],
        errors="coerce"
    ).fillna(0)

    # Приведение площади к числовому формату
    d["size"] = pd.to_numeric(d["size"], errors="coerce")

    # Дополнительные признаки
    d["size_per_room"] = d["size"] / (d["rooms"] + 1)
    d["is_new_building"] = (d["building_age_num"] <= 5).astype(int)
    d["floor_ratio"] = (
        d["floor_num"] / (d["total_floors_num"] + 1)
    )
    d["has_living_room"] = (d["living_rooms"] > 0).astype(int)

    # Объединение редких категорий
    subtype = d["sub_type"].fillna("Unknown")
    heating = d["heating_type"].fillna("Unknown")

    if top_subtypes is not None:
        d["sub_type_grouped"] = subtype.where(
            subtype.isin(top_subtypes), "Other"
        )
    else:
        d["sub_type_grouped"] = subtype

    if top_heating is not None:
        d["heating_grouped"] = heating.where(
            heating.isin(top_heating), "Other"
        )
    else:
        d["heating_grouped"] = heating

    # Выбираем признаки для модели
    feature_columns = [
        "size",
        "rooms",
        "living_rooms",
        "building_age_num",
        "total_floors_num",
        "is_penthouse",
        "is_garden",
        "is_ground",
        "floor_num",
        "size_per_room",
        "is_new_building",
        "floor_ratio",
        "has_living_room",
        "sub_type_grouped",
        "heating_grouped",
        "price_currency",
        "city"
    ]

    features = d[feature_columns].copy()

    # Преобразование категорий в числовые столбцы
    features = pd.get_dummies(
        features,
        columns=[
            "sub_type_grouped",
            "heating_grouped",
            "price_currency",
            "city"
        ],
        drop_first=True,
        dtype=int
    )

    return features
```

## 4.4. Подготовка признаков для обучения и тестирования

Для корректного преобразования категориальных признаков сначала определяются наиболее частые типы недвижимости и системы отопления в обучающей выборке.

После этого признаки формируются отдельно для обучающих объектов и тестовых объектов продажи. Столбцы выравниваются, чтобы обучающая и тестовая матрицы имели одинаковый набор признаков.

Пропущенные числовые значения заполняются медианой. Затем применяется RobustScaler, который использует медиану и межквартильный размах для масштабирования признаков.

```python
# Наиболее частые категории определяются только по обучающим данным
top_subtypes = (
    train["sub_type"]
    .fillna("Unknown")
    .value_counts()
    .head(5)
    .index
)

top_heating = (
    train["heating_type"]
    .fillna("Unknown")
    .value_counts()
    .head(6)
    .index
)

# Из тестового файла выбираем только объявления о продаже
sale_test = test_samples[
    test_samples["listing_type"] == 1
].copy()

# Формируем признаки
X_train = make_features(train, top_subtypes, top_heating)
X_test = make_features(sale_test, top_subtypes, top_heating)

# Согласуем наборы столбцов
X_train, X_test = X_train.align(
    X_test,
    join="left",
    axis=1,
    fill_value=0
)

# Заполнение пропусков
imputer = SimpleImputer(strategy="median")

X_train_i = imputer.fit_transform(X_train)
X_test_i = imputer.transform(X_test)

# Масштабирование признаков
scaler = RobustScaler()

X_train_s = scaler.fit_transform(X_train_i)
X_test_s = scaler.transform(X_test_i)

print("Обучающая матрица:", X_train_s.shape)
print("Тестовая матрица:", X_test_s.shape)
print("Количество тестовых объявлений о продаже:", len(sale_test))
```

## 4.5. Обучение модели Random Forest

Для прогнозирования используется алгоритм Random Forest — ансамбль деревьев решений. Он способен учитывать нелинейные зависимости между характеристиками недвижимости и её стоимостью.

Чтобы результаты можно было воспроизвести, задаётся фиксированное значение `random_state=42`. Для обучения используется случайная выборка размером до 20 000 объектов из подготовленных исходных данных.

Модель обучается на известных ценах продажи, после чего используется для прогнозирования цен тестовых объектов.

```python
# Целевая переменная — фактическая цена продажи
y_train = train["price"].to_numpy(dtype=float)

# Случайный выбор до 20 000 объектов для ускорения обучения
rng = np.random.RandomState(42)

chosen = rng.choice(
    len(X_train_s),
    size=min(20000, len(X_train_s)),
    replace=False
)

# Создание и обучение модели
model = RandomForestRegressor(
    n_estimators=25,
    max_depth=12,
    min_samples_leaf=3,
    random_state=42,
    n_jobs=1
)

model.fit(
    X_train_s[chosen],
    y_train[chosen]
)

# Прогнозы для тестовых объектов продажи
sale_predictions = model.predict(X_test_s)

print("Модель обучена.")
print("Количество объектов, использованных для обучения:", len(chosen))
print("Количество рассчитанных прогнозов:", len(sale_predictions))
```

## 4.7. Расчёт метрик качества вручную

Для проверки качества модели используются фактические цены тестовых объектов продажи и рассчитанные моделью прогнозы.

MAE — средняя абсолютная ошибка между фактическими и прогнозируемыми ценами. Она показывает средний размер ошибки в турецких лирах.

RMSE — корень из средней квадратичной ошибки. Эта метрика сильнее реагирует на большие отклонения прогнозов от фактических значений.

R² — коэффициент детерминации, который характеризует качество прогнозирования относительно базового предсказания средней цены.

Сначала метрики рассчитываются вручную с помощью NumPy. Затем полученные значения сравниваются с результатами функций библиотеки sklearn.

```python
# Фактические цены и прогнозы для тестовых продаж
y_true = sale_test["price"].to_numpy(dtype=float)
y_pred = sale_predictions

# MAE: средняя абсолютная ошибка
mae_manual = np.mean(np.abs(y_true - y_pred))

# RMSE: корень из средней квадратичной ошибки
rmse_manual = np.sqrt(np.mean((y_true - y_pred) ** 2))

# R²: коэффициент детерминации
ss_res = np.sum((y_true - y_pred) ** 2)
ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)

r2_manual = (
    1 - ss_res / ss_tot
    if ss_tot != 0
    else np.nan
)

print("MAE, ручной расчёт:", round(mae_manual, 4))
print("RMSE, ручной расчёт:", round(rmse_manual, 4))
print("R², ручной расчёт:", round(r2_manual, 4))
```

## 4.8. Проверка метрик с помощью sklearn

Для дополнительной проверки рассчитываются те же метрики с использованием готовых функций sklearn.

Сравнение двух способов расчёта позволяет проверить правильность реализованных вручную формул. Значения должны совпадать с учётом возможного округления.

```python
mae_sklearn = mean_absolute_error(y_true, y_pred)

rmse_sklearn = np.sqrt(
    mean_squared_error(y_true, y_pred)
)

r2_sklearn = r2_score(y_true, y_pred)

metrics_comparison = pd.DataFrame({
    "Метрика": ["MAE", "RMSE", "R²"],
    "Ручной расчёт": [
        mae_manual,
        rmse_manual,
        r2_manual
    ],
    "Расчёт sklearn": [
        mae_sklearn,
        rmse_sklearn,
        r2_sklearn
    ]
})

display(metrics_comparison.round(4))
```
